### Placement Prediction

Introduction

In this project, we aim to predict whether a student will secure a job placement based on various academic and demographic attributes, including CGPA, soft skills rating, and projects number.

University graduates often face challenges in securing employment, and understanding the factors that influence placement success can guide educational institutions and students in improving outcomes. By applying machine learning techniques to historical placement data, we can build predictive models that assist in identifying students at risk of not being placed and recommend targeted interventions.

We will perform the following steps in this notebook:

- Load and preprocess the dataset
- Explore and visualize the data
- Train and evaluate classification models


Let’s get started

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Data = pd.read_csv('placementdata.csv')
Data.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
elements = ['ExtracurricularActivities','PlacementTraining','PlacementStatus']
for elem in elements :
  Data[elem] = encoder.fit_transform(Data[elem])
```

```python
Data.info()
```

```python
Data.drop(columns=['StudentID'],inplace=True)
```

```python
Data.isnull().sum().sum()

# check for missing values  
```

```python
Data.duplicated().sum()

# check for duplicates values
```

```python
Data.drop_duplicates(inplace=True)
```

```python
x = Data.drop(columns=['PlacementStatus'])   
y = Data['PlacementStatus']
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

Model = LogisticRegression()
Model.fit(x_train,y_train)
```

```python
y_pred = Model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test,y_pred)
print(f"""The Accuracy : {accuracy * 100:.2f}%""")

# evaluate the model with accuracy score
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,y_pred))

# evaluate the model with classification report
```

```python
print(f"Train Score : {Model.score(x_train, y_train) * 100:.2f}%")
print(f"Test Score  : {Model.score(x_test, y_test) * 100:.2f}%")

# display training and test accuracy
```
